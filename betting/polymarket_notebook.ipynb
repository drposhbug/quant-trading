```python
import time as time_module

def fetch_price_history_best_effort(token_id):
    """Try progressively coarser fidelity until something returns data."""
    for fidelity in [60, 360, 720]:  # 1hr, 6hr, 12hr
        r = requests.get("https://clob.polymarket.com/prices-history", params={
            "market": token_id, "interval": "max", "fidelity": fidelity
        })
        if r.status_code == 200:
            data = r.json().get("history", [])
            if data:
                return pd.DataFrame(data), fidelity
    return pd.DataFrame(), None

results = []
for _, m in moneyline_df.iterrows():
    try:
        token_ids = json.loads(m["clobTokenIds"])
        outcome_prices = json.loads(m["outcomePrices"])
        prices, fidelity_used = fetch_price_history_best_effort(token_ids[0])
        results.append({
            "question": m["question"],
            "final_price": outcome_prices[0],
            "ticks": len(prices),
            "fidelity_used": fidelity_used
        })
    except Exception as e:
        results.append({"question": m["question"], "ticks": 0, "error": str(e)[:50]})
    time_module.sleep(0.15)

history_check_df = pd.DataFrame(results)
print(f"Markets with SOME price history: {(history_check_df['ticks'] > 0).sum()} / {len(history_check_df)}")
history_check_df[history_check_df["ticks"] > 0].head(10)
```
